```python
invalidos = df_prf_original['pessoas'] < 0
print("Número de pessoas negativo:", invalidos.sum())

invalidos = df_prf_original['mortos'] < 0
print("Número de mortos:", invalidos.sum())

invalidos = df_prf_original['feridos_leves'] < 0
print("Número de feridos leves negativos:", invalidos.sum())

invalidos = df_prf_original['feridos_graves'] < 0
print("Número de feridos graves negativo:", invalidos.sum())

invalidos = df_prf_original['ilesos'] < 0
print("Número de ilesos negativo:", invalidos.sum())

invalidos = df_prf_original['feridos'] < 0
print("Número de feridos negativo:", invalidos.sum())
```
